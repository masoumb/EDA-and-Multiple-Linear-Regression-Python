### Task 1 : Import packages and data loading


```python
import pandas as pd
from scipy import stats
```

```python
taxi_data = pd.read_csv("2017_Yellow_Taxi_Trip_Data.csv", index_col = 0)
```

### Task 2 : Data exploration
Exploratory Data Analysis for thid dataset has already been performed in another notebook.

```python
# Descriptive statistics
taxi_data.describe(include='all')
```

**Note:** In the dataset, `payment_type` is encoded in integers:
*   1: Credit card
*   2: Cash
*   3: No charge
*   4: Dispute
*   5: Unknown

- Since we are interested in the relationship between payment type and fare amount, we can look into the fare amount per each category 0f payment type. 

```python
group_payment_fare = taxi_data[['payment_type', 'fare_amount']].groupby('payment_type').mean('fare_amount').reset_index()
group_payment_fare
```

- As we can see, the mean fare amount of customers who paid with their crredit card is higher than the average fare amount of other type payments.
- However, the observed difference might be due to the random sampling and chance. In order to evaluate that if the difference is statistically significant, we can perform a hypothesis testing.  

### Task 3 : Hypothesis testing 

- The goal here is to conduct a two-sample t-test following the next steps:
1.   State the null hypothesis and the alternative hypothesis
2.   Choose a signficance level
3.   Find the p-value
4.   Reject or fail to reject the null hypothesis 

**1. State the null and alternative hypothesis:**

**Null Hypothesis ($H_0$):** 
    The difference in the observed average of fare amount for different payment types is due to the chance. And there is no difference in the average of fare amounts observed for customers paying with credit cards and those who pay with cash.
    
**Alternative Hypothesis ($H_A$):**
    The difference in the observed average of fare amount for different payment types is statistically significant. There is a difference in the average fare amount between customers who pay with cedit car and others who use cash.

**2. Choose a significance level:**
    
   Here we use 5% significance level.

**3. Find the p-value:**

```python
credit_data = taxi_data[taxi_data['payment_type']==1]['fare_amount']
cash_data = taxi_data[taxi_data['payment_type']==2]['fare_amount']
```

```python
stats.ttest_ind(a=credit_data, b=cash_data, equal_var=False)
```

**4. Reject or fail to reject the null hypothesis:**

Since the p-value=6.8e-12 is less than 0.05, we reject the null hypothesis. Therefore, we can say that there is a statistically significant difference in the average fare amount between customers who use credit cards and customers who use cash.
